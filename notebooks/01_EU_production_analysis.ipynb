```python
"1. Which countries have seen increased production of refined petrolium products since the begining of war?"

# 1. Defines the specific desired output (what refineries actually produce) 
refined_mask = (
    (df['nrg_bal'] == 'TO_RPI_RO') & 
    (df['siec'] == 'O4000XBIO') & 
    (df['unit'] == 'THS_T') &
    (df['2021'] > 0) & (df['2024'])
)

# 2. Isolate countries and apply the filter
country_df = df[refined_mask].copy()

# 3. Calculate the Growth
country_df['absolute_increase'] = country_df['2024'] - country_df['2021']
country_df['pct_growth'] = (country_df['absolute_increase'] / country_df['2023']) * 100

# 4. Final Result
top_performers = country_df[['geo\\TIME_PERIOD', '2021', '2024', 'absolute_increase', 'pct_growth']].sort_values(by='absolute_increase', ascending=False)

print("--- Cleaned Refinery Output Growth (2021 - 2024) ---")
print(top_performers.head(10))
```

```python
"2. Which countries have seen a decreased production of refined petrolium products since the begining of war?"

# 1. Defines the specific desired output (what refineries actually produce) 
refined_mask = (
    (df['nrg_bal'] == 'TO_RPI_RO') & 
    (df['siec'] == 'O4000XBIO') & 
    (df['unit'] == 'THS_T') &
    (df['2021'] > 0) & (df['2024'] > 0) 
)

# 2. Isolate countries and apply the filter
country_df = df[refined_mask].copy()
country_df = country_df[country_df['geo\\TIME_PERIOD'].str.len() == 2] # str.len() is used to remove special aggregates like EU20

# 3. Calculate the Contraction
country_df['absolute_decrease'] = country_df['2024'] - country_df['2021']
country_df['pct_contraction'] = (country_df['absolute_decrease'] / country_df['2021']) * 100

# 4. Final Result
top_performers = country_df[['geo\\TIME_PERIOD', '2021', '2024', 'absolute_decrease', 'pct_contraction']].sort_values(by='absolute_decrease', ascending=True)

print("--- Cleaned Refinery Output Decrease (2021 - 2024) ---")
print(top_performers.head(10))

#note for analysis: there was a slower decrease in production between 2021 and 2023 followed by a contraction
```

```python
"3. Has EU production alone been sufficient to offset the Russian decline, or does a gap remain?"

# 1. Filter for total EU Production and Imports
mask_ppr_imp = (
    (df['geo\\TIME_PERIOD'] == 'EU27_2020') & 
    (df['siec'] == 'O4000XBIO') & 
    (df['nrg_bal'].isin(['PPR', 'IMP'])) & # PPR = Production, IMP = Imports
    (df['unit'] == 'THS_T')
)

# 2. Filter to see the comparison between 2021 and 2024
comparison_df = df[mask_ppr_imp][['nrg_bal', '2021', '2022', '2023', '2024']]

# 3. Result
print(comparison_df)
```
