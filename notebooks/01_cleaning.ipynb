```python
"1. Which countries has increased production of refined petrolium products since the begining of war?"

# 1. Defines the specific desired output (what refineries actually produce) 
refined_mask = (
    (df['nrg_bal'] == 'TO_RPI_RO') & 
    (df['siec'] == 'O4000XBIO') & 
    (df['unit'] == 'THS_T') &
    (df['2023'] > 0) & (df['2024'] > 0)
)

# 2. Isolate countries and apply the filter
country_df = df[refined_mask].copy()
country_df = country_df[country_df['geo\\TIME_PERIOD'].str.len() == 2] # str.len() is used to remove special aggregates like EU20

# 3. Calculate the Growth
country_df['absolute_increase'] = country_df['2024'] - country_df['2023']
country_df['pct_growth'] = (country_df['absolute_increase'] / country_df['2023']) * 100

# 4. Final Result
top_performers = country_df[['geo\\TIME_PERIOD', '2023', '2024', 'absolute_increase', 'pct_growth']].sort_values(by='absolute_increase', ascending=False)

print("--- Cleaned Refinery Output Growth (2023 vs 2024) ---")
print(top_performers.head(10))
```
